0001 - Tính chu vi và diện tích hình vuông

Đề bài

Viết chương trình tính chu vi và diện tích hình vuông, với cạnh hình vuông a là số
nguyên theo công thức cv=a*4, dt=a*a

Dữ liệu vào

Số nguyên dương a là cạnh hình vuông

Dữ liệu ra

Chu vi và diện tích nằm trên cùng một dòng, cách nhau bởi dấu cách

```python
a = int(input())
cv = a*4
dt = a*a
print (cv, dt)
```

0002 - Tính chu vi và diện tích hình tròn

Đề bài

Viết chương trình tính chu vi và diện tích hình tròn, với bán kính r là số nguyên theo
công thức cv=r*2*3.14, dt=r*r*3.14.
Dữ liệu vào

Một số nguyên dương r là bán kính hình tròn.

Dữ liệu ra

Là chu vi và diện tích của hình tròn nằm trên cùng một dòng, cách nhau bởi dấu cách,
làm tròn đến 2 chữ số thập phân.


```python
r = int(input())
cv = r * 2 * 3.14
dt = r * r * 3.14
print (cv, dt)
```

0004 - Tính tổng và tích 2 số nguyên a và b

Đề bài

Viết chương trình tính tổng và tích 2 số nguyên a và b

Dữ liệu vào

Một dòng có 2 số nguyên cách nhau bởi dấu cách theo thứ tự là a và b

Dữ liệu ra

Tổng và tích nằm trên cùng một dòng, cách nhau bởi dấu cách

```python
a , b = map(int, input().split())
tong = a + b
tich = a * b
print (tong, tich)
```
