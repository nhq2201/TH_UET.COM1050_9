0006 - Tính chu vi và diện tích của một tam giác

Đề bài
Viết chương trình tính chu vi (cv) và diện tích (s) một tam giác có 3 cạnh a, b, c là các
số nguyên theo công thức cv=a+b+c, s=sqrt(p*(p-a)*(p-b)*(p-c)) với p=cv/2 (3 số a, b, c
bảo đảm là 3 cạnh của 1 tam giác).

Dữ liệu vào

Gồm 1 dòng có 3 số nguyên cách nhau bởi dấu cách theo thứ tự là a, b và c.

Dữ liệu ra

Là chu vi và diện tích nằm trên cùng một dòng, cách nhau bởi dấu cách, chu vi làm tròn
đến 1 chữ số thập phân, diện tích làm tròn đến 3 chữ số thập phân.


```python
a , b , c = map(int, input().split())
cv = a + b + c
p = (a + b + c)/2
s = (p * (p - a) * (p - b) * (p - c))**0.5
if (a+b)>c and (b+c)> a and (c+a)>b and a>0 and b>0 and c>0:
    print (f"{cv:.1f}", f"{s:.3f}")
else:
    raise ValueError("Khong phai la tam giac")
```

0008 - Tính bán kính R của đường tròn ngoại tiếp tam giác

là các số nguyên theo công thức R=abc/4S , với S là diện tích tam giác (3 số a, b, c bảo
đảm là 3 cạnh của 1 tam giác).

Dữ liệu vào

Một dòng có 3 số nguyên cách nhau bởi dấu cách theo thứ tự là a, b và c (đảm bảo là 3
cạnh của một tam giác).

0015 - Kiểm tra a, b, c có là 3 cạnh của một tam giác
Đề bài
Viết chương trình kiểm tra xem 3 số a, b, c nhập từ bàn phím có là 3 cạnh của một tam
giác hay ko? Nếu phải thì tính chu vi và diện tích của tam giác đó, ngược lại thông báo
3 số đó ko phải là 3 cạnh của tam giác

Dữ liệu vào

Một dòng có 3 số nguyên cách nhau bởi dấu cách theo thứ tự là a, b và c.

Dữ liệu ra

Dòng 1: “Day khong phai la 3 canh cua mot tam giac” hoặc “Day la 3 canh cua mot tam
giac”, nếu là trường hợp 2 thì thêm dòng 2 là chu vi và diện tích nằm trên cùng một
dòng, cách nhau bởi dấu cách, chu vi làm tròn đến 2 chữ số thập phân, diện tích làm
tròn đến 1 chữ số thập phân.

```python
a, b, c = map(int, input().split())
cv = a + b + c
p = (a + b + c)/2
s = (p * (p - a) * (p - b) * (p - c))**0.5
if (a+b)>c and (b+c)> a and (c+a)>b and a>0 and b>0 and c>0:
    print ("Day la 3 canh cua mot tam giac")
    print (f"{cv:.2f}" + " " + f"{s:.1f}")
else:
    print ("Day khong phai la 3 canh cua mot tam giac")

```

0016 - Xếp loại học sinh

Đề bài

Viết chương trình nhập vào điểm trung bình học kỳ I và trung bình học kỳ II, tính trung
bình cả năm theo công thức: TBCN=(TBHKI+TBHKII*2)/3. Xếp lọai theo cách:

“Gioi” nếu TBCN>=8

“Kha” nếu 6.5<=TBCN <8

“Trung binh” nếu 5.0<=TBCN <6.5

“Yeu” nếu 3.5<=TBCN <5

“Kem” nếu TBCN<3.5

Dữ liệu vào

Một dòng có 2 số thực cách nhau bởi dấu cách theo thứ tự là TBHKI và TBHKII

Dữ liệu ra

Dòng đầu là TBCN, làm tròn đến 1 chữ số thập phân; dòng thứ hai là xếp loại

```python
a, b = map(int, input().split())
c = (a+ b*2)/3
if c < 3.5:
    print (f"{c:.1f}")
    print ("Kem")
elif c < 5:
    print (f"{c:.1f}")
    print ("Yeu")
elif c < 6.5:
    print (f"{c:.1f}")
    print ("Trung binh")
elif c < 8:
    print (f"{c:.1f}")
    print ("Kha")
else:
    print (f"{c:.1f}")
    print ("Gioi")

```
